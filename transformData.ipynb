```python
import pandas as pd
```

```python
def process_weather_data(file_path):
    flight_data = pd.read_csv(file_path)
    # Convert the timestamp to a pandas datetime object
    datetime_obj = pd.to_datetime(flight_data['timestamp_local'])

    # Extract the hour and minute components
    flight_data['hour'] = datetime_obj.dt.hour
    flight_data['minutes'] = datetime_obj.dt.minute

    columns_to_drop = ['app_temp','azimuth', 'dhi', 'dni', 'elev_angle', 'ghi', 'pod', 'revision_status', 'slp', 'solar_rad', 'timestamp_local', 'timestamp_utc', 'ts', 'uv', 'weather.icon']
    flight_data.drop(columns=columns_to_drop, inplace=True)
    
    return flight_data
```

```python
# List of file paths
file_paths = ['JFK_weather_data_sub_hourly.csv', 'MCO_weather_data_sub_hourly.csv', 
              'ORD_weather_data_sub_hourly.csv', 'SYR_weather_data_sub_hourly.csv']

for file in file_paths:
    processed_data = process_weather_data(file)
    processed_data.head()
    print(len(processed_data.columns))
```
